## Sequential Design Pattern

#### Import libraries

```python
from agents import Agent, Runner, trace
import os, asyncio
from dotenv import load_dotenv

load_dotenv(override=True)
```

#### Define Agents

```python
# Agent 1: Extract gene names
extract_genes_agent = Agent(
    name="GeneExtractor",
    instructions="Extract all gene names mentioned in the given biomedical text. Return as a comma-separated list.",
    model="gpt-4.1-nano"
)

# Agent 2: Annotate gene functions
annotate_genes_agent = Agent(
    name="GeneAnnotator",
    instructions="Given a list of gene names, provide a brief biological function for each gene in just 15 words or less.",
    model="gpt-4.1-nano"
)
```

#### Define Sequential workflow

```python
async def sequential_pipeline(text):
    # Step 1: Extract gene names
    result1 = await Runner.run(extract_genes_agent, text)
    gene_list = result1.final_output
    print(f"Extracted Genes: {gene_list}")

    # Step 2: Annotate gene functions
    # await asyncio.sleep(30)
    result2 = await Runner.run(annotate_genes_agent, gene_list)
    return result2.final_output
```

#### Execute the workflow

```python
# Example usage
with trace("sequential_pipeline_demo"):
    text = "BRCA1 and TP53 are important genes in cancer biology. EGFR is also frequently mutated."
    output = await sequential_pipeline(text)
    print(output)
```
