```python
# tooluniverse-mcp --help
from agents.mcp import MCPServerStdio

fetch_params = {"command": "tooluniverse-smcp-stdio", "args": ["--categories", "uniprot", "ChEMBL", "opentarget", "--hooks", "--hook-type", "SummarizationHook"]}

async with MCPServerStdio(params=fetch_params, client_session_timeout_seconds=300) as server:
    fetch_tools = await server.list_tools()

fetch_tools
```

```python
for tool in fetch_tools:
    print(tool.name)
```

```python
# List UniProt-related tools from fetch_tools
uniprot_tools = [tool for tool in fetch_tools if tool.name.startswith("UniProt")]
for tool in uniprot_tools:
    print(tool.name, "-", tool.description)
```

```python
instructions_uniprot = "You are an expert bioinformatics agent specialized in retrieving and summarizing information from the UniProt database. You have access to mcp server tools that starts with UniProt_. Use these tools to answer queries related to UniProt"

from agents.mcp import MCPServerStdio, create_static_tool_filter

# Only allow tools whose names start with "uniprot_"
tool_filter = create_static_tool_filter(allowed_tool_names=[ "UniProt_get_function_by_accession"])

async with MCPServerStdio(params=fetch_params, client_session_timeout_seconds=300, tool_filter=tool_filter) as server:
    uniprot_agent = Agent(name="uniprot_agent",
                    instructions=instructions_uniprot,
                    model="gpt-4.1-nano",
                    mcp_servers=[server])
    
    query="Tell me about the function of protein uniprot id P04637"
    result = await Runner.run(uniprot_agent, query)
    print(result.final_output)
   
```
